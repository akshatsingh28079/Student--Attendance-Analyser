```python
import pandas as pd
import numpy as np
```

```python
# total classes held for each subject
CLASSES_PER_SUB = 30
SUBJECTS = ['Maths', 'Python', 'Physics', 'English']
TOTAL_CLASSES = CLASSES_PER_SUB * len(SUBJECTS)

# sample student data
data = {
    'Roll_No': [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],
    'Name': ['Amay', 'Arun', 'Alekh', 'Akshat', 'Aditya', 'Anvit', 'Jatin Parashar', 'Jatin Tyagi', 'Ishant', 'Aryan'],
    'Maths': [26, 28, 18, 25, 22, 29, 15, 24, 20, 27],
    'Python': [28, 27, 20, 26, 24, 30, 19, 25, 21, 28],
    'Physics': [24, 26, 17, 23, 19, 27, 16, 22, 18, 25],
    'English': [25, 29, 21, 24, 23, 28, 20, 26, 22, 26]
}

df = pd.DataFrame(data)
```

```python
def calculate_attendance(dataframe):
    # total classes attended across all subjects
    dataframe['Total_Attended'] = dataframe[SUBJECTS].sum(axis=1)

    # percentage calculation using numpy
    pct = (dataframe['Total_Attended'] / TOTAL_CLASSES) * 100
    dataframe['Percentage'] = np.round(pct, 2)

    # 75% criteria for exam eligibility
    dataframe['Status'] = np.where(dataframe['Percentage'] >= 75, 'Eligible', 'Detained')
    return dataframe
```

```python
def show_all_records(dataframe):
    print("\n--- All Student Attendance Records ---")
    print(dataframe.to_string(index=False))
```

```python
def show_statistics(dataframe):
    pct_arr = np.array(dataframe['Percentage'])

    print("\n--- Class Attendance Statistics ---")
    print("Total Students       :", len(dataframe))
    print(f"Average Attendance   : {np.mean(pct_arr):.2f}%")
    print(f"Highest Attendance   : {np.max(pct_arr):.2f}%")
    print(f"Lowest Attendance    : {np.min(pct_arr):.2f}%")
    print(f"Standard Deviation   : {np.std(pct_arr):.2f}")

    eligible_count = int(np.sum(pct_arr >= 75))
    detained_count = int(np.sum(pct_arr < 75))
    print("Eligible (>= 75%)    :", eligible_count)
    print("Detained (< 75%)     :", detained_count)

    # average attendance per subject
    print(f"\nSubject-wise Average Attendance (out of {CLASSES_PER_SUB} classes):")
    for sub in SUBJECTS:
        avg_sub = np.mean(dataframe[sub])
        print(f"  {sub:8s} : {avg_sub:.2f}")
```

```python
def show_defaulters(dataframe):
    low_att = dataframe[dataframe['Percentage'] < 75]
    print("\n--- Defaulter List (Below 75%) ---")
    if low_att.empty:
        print("No students are below 75% attendance.")
    else:
        print(low_att[['Roll_No', 'Name', 'Total_Attended', 'Percentage', 'Status']].to_string(index=False))
```

```python
def search_student(dataframe):
    try:
        roll = int(input("Enter Roll Number to search: "))
    except ValueError:
        print("Please enter a valid number.")
        return

    result = dataframe[dataframe['Roll_No'] == roll]
    if result.empty:
        print("No record found for Roll No:", roll)
    else:
        print("\n--- Student Record ---")
        print(result.to_string(index=False))
```

```python
def add_student(dataframe):
    try:
        roll = int(input("Enter Roll Number: "))
        if roll in dataframe['Roll_No'].values:
            print("This Roll Number already exists.")
            return None

        name = input("Enter Student Name: ").strip()
        new_data = {'Roll_No': roll, 'Name': name}

        for sub in SUBJECTS:
            att = int(input(f"Enter classes attended in {sub} (out of {CLASSES_PER_SUB}): "))
            if att < 0 or att > CLASSES_PER_SUB:
                print(f"Attendance must be between 0 and {CLASSES_PER_SUB}.")
                return None
            new_data[sub] = att
    except ValueError:
        print("Invalid input. Enter numbers for Roll No and classes.")
        return None

    new_row = pd.DataFrame([new_data])
    dataframe = pd.concat([dataframe, new_row], ignore_index=True)
    dataframe = calculate_attendance(dataframe)
    print("Student record added.")
    return dataframe
```

```python
def save_report(dataframe):
    filename = "attendance_report.csv"
    dataframe.to_csv(filename, index=False)
    print("Saved report to", filename)
```

```python
# run initial calculation before showing menu
df = calculate_attendance(df)
```

```python
# main menu
while True:
    print("\n==============================")
    print("  STUDENT ATTENDANCE ANALYSER ")
    print("==============================")
    print("1. View All Records")
    print("2. View Class Statistics")
    print("3. View Defaulter List (< 75%)")
    print("4. Search Student by Roll No")
    print("5. Add New Student")
    print("6. Export Report to CSV")
    print("7. Exit")

    choice = input("Enter your choice (1-7): ")

    if choice == '1':
        show_all_records(df)
    elif choice == '2':
        show_statistics(df)
    elif choice == '3':
        show_defaulters(df)
    elif choice == '4':
        search_student(df)
    elif choice == '5':
        updated_df = add_student(df)
        if updated_df is not None:
            df = updated_df
    elif choice == '6':
        save_report(df)
    elif choice == '7':
        print("Exiting program.")
        break
    else:
        print("Invalid choice, try again.")
```
